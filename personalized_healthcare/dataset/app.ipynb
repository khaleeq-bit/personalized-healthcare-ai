```python
import pickle
import pandas as pd

# Load trained model
with open("disease_model.pkl", "rb") as file:
    model = pickle.load(file)

# Load symptom encoder
with open("symptom_encoder.pkl", "rb") as file:
    mlb = pickle.load(file)

# Load feature columns
with open("feature_columns.pkl", "rb") as file:
    feature_columns = pickle.load(file)

print("Model loaded successfully!")
print("Symptom encoder loaded successfully!")
print("Feature columns loaded successfully!")
```

```python
# Cell 5: Prediction Function

def predict_disease(age, symptoms):
    
    # Create an empty row for all symptoms
    symptom_vector = pd.DataFrame(
        0,
        index=[0],
        columns=mlb.classes_
    )
    
    # Convert symptoms to lowercase
    symptoms = [symptom.strip().lower() for symptom in symptoms]
    
    # Set selected symptoms to 1
    for symptom in symptoms:
        if symptom in symptom_vector.columns:
            symptom_vector.loc[0, symptom] = 1
    
    # Add age
    symptom_vector["Age"] = age
    
    # Add symptom count
    symptom_vector["Symptom_Count"] = len(symptoms)
    
    # Arrange columns exactly like the training data
    symptom_vector = symptom_vector[feature_columns]
    
    # Predict
    prediction = model.predict(symptom_vector)[0]
    
    return prediction


print("Prediction function created successfully!")
```

```python
# Cell 6: Test Prediction

age = 25

symptoms = [
    "fever",
    "cough",
    "headache"
]

result = predict_disease(age, symptoms)

print("Age:", age)
print("Symptoms:", ", ".join(symptoms))
print("Predicted Condition:", result)
```

```python
# Cell 7: Flask API

from flask import Flask, request, jsonify

app = Flask(__name__)


@app.route("/predict", methods=["POST"])
def predict():

    # Get data sent from the website
    data = request.get_json()

    # Get age
    age = int(data["age"])

    # Get symptoms
    symptoms = data["symptoms"]

    # Use our ML prediction function
    prediction = predict_disease(age, symptoms)

    # Send prediction back to the website
    return jsonify({
        "prediction": prediction
    })


print("Flask API created successfully!")
```

```python
# Cell 8: Home Page Route

from flask import render_template

@app.route("/")
def home():
    return render_template("index.html")


print("Home page route created successfully!")
```
